```python
# Q1 — Clean and validate a city name
city = ' nEw dELhi '
clean_city = city.strip().title()
valid_city = True
for character in clean_city:
    if not (character.isalpha() or character == ' '):
        valid_city = False
        break

if valid_city:
    print(clean_city)
else:
    print('Invalid city')
```

```python
# Q2 — Convert prices and calculate the bill
price_text = ' 120, 45 , 300 ,80, 95 '
prices = []
for price in price_text.split(','):
    prices.append(int(price.strip()))

total = sum(prices)
highest = max(prices)
average = total / len(prices)
print(f'Total : {total}')
print(f'Highest : {highest}')
print(f'Average : {average:.2f}')
```

```python
# Q3 — Validate a username
def validate_username(username):
    username = username.strip().lower()
    if (username.isalnum() and username[0].isalpha() and
            5 <= len(username) <= 12):
        return 'Valid username'
    return 'Invalid username'

print(validate_username(' Rahul_123 '))
print(validate_username(' rahul123 '))
```

```python
# Q4 — Count words, vowel-ending words, and find the longest word
sentence = 'Python makes data analysis interesting'
words = sentence.split()
vowel_ending_count = 0
for word in words:
    if word[-1].lower() in 'aeiou':
        vowel_ending_count += 1

longest_word = words[0]
for word in words:
    if len(word) > len(longest_word):
        longest_word = word

print(f'Word count : {len(words)}')
print(f'Words ending with a vowel : {vowel_ending_count}')
print(f'Longest word : {longest_word}')
```

```python
# Q5 — Character frequency, ignoring spaces and case
text = 'Aba c'
character_frequency = {}
for character in text.lower():
    if character != ' ':
        character_frequency[character] = character_frequency.get(character, 0) + 1

print(character_frequency)
```

```python
# Q6 — Most frequent word
sentence = 'red blue red green blue red'
word_frequency = {}
for word in sentence.lower().split():
    word_frequency[word] = word_frequency.get(word, 0) + 1

sorted_words = sorted(word_frequency.items(), key=lambda item: item[1], reverse=True)
print(f'Most frequent word : {sorted_words[0][0]}')
print(f'Frequency : {sorted_words[0][1]}')
```

```python
# Q7 — Invert a dictionary while grouping duplicate values
sales = {'mon': 200, 'tue': 350, 'wed': 200}
inverted_sales = {}
for day, amount in sales.items():
    if amount not in inverted_sales:
        inverted_sales[amount] = []
    inverted_sales[amount].append(day)

print(inverted_sales)
```

```python
# Q8 — Merge two test dictionaries and sum overlapping students
test1 = {'amit': 40, 'bina': 35}
test2 = {'amit': 30, 'chirag': 45}
combined_marks = test1.copy()
for student, marks in test2.items():
    combined_marks[student] = combined_marks.get(student, 0) + marks

print(combined_marks)
```

```python
# Q9 — Sort employees by salary, highest first
employees = [('Ravi', 55000), ('Sita', 72000), ('Amit', 48000)]
sorted_employees = sorted(employees, key=lambda employee: employee[1], reverse=True)
for name, salary in sorted_employees:
    print(f'{name} earns Rs. {salary}')
```

```python
# Q10 — Inspect JSON-style movie data
movies = [
    {'title': 'Movie A', 'rating': 8.2, 'genre': 'Thriller'},
    {'title': 'Movie B', 'rating': 8.4, 'genre': 'Sports'},
    {'title': 'Movie C', 'rating': 8.1, 'genre': 'Comedy'},
    {'title': 'Movie D', 'rating': 8.2, 'genre': 'Drama'},
]

print('(a) Movies with rating > 8.1:')
for movie in movies:
    if movie['rating'] > 8.1:
        print(movie['title'])

genres = []
for movie in movies:
    if movie['genre'] not in genres:
        genres.append(movie['genre'])
print('(b) Genres:', genres)

movies_by_rating = {}
for movie in movies:
    rating = movie['rating']
    if rating not in movies_by_rating:
        movies_by_rating[rating] = []
    movies_by_rating[rating].append(movie['title'])

print('(c) Movies grouped by rating:')
for rating in sorted(movies_by_rating):
    titles = ', '.join(movies_by_rating[rating])
    print(f'Rating {rating:.1f}: {titles}')
```
