## Q5. Scholarship Shortlisting

```python
totals = np.array([245, 310, 178, 392, 140, 265, 199, 155, 288, 132])

# Task 1: Totals of students who scored above 250
print("Task 1:", totals[totals > 250])

# Task 2: Totals scoring above 180 but below 250
band = totals[(totals > 180) & (totals < 250)]
print("\nTask 2:", band)   # Expected: [245 199]

# Task 3: Totals either below 150 or above 380
print("\nTask 3:", totals[(totals < 150) | (totals > 380)])

# Task 4: Count in the band of task 2
print("\nTask 4:", np.sum((totals > 180) & (totals < 250)))

# Task 5: Average of students who scored above 200
print("\nTask 5:", totals[totals > 200].mean())

# Task 6: Index of every student below 150
print("\nTask 6:", np.where(totals < 150)[0])   # Expected: [4 9]

# Task 7: Highest total among students NOT in merit band (above 250)
print("\nTask 7:", totals[totals <= 250].max())
```

## Q6. After Revaluation

```python
marks = np.array([40, 55, 62, 88, 95, 33, 71, 58])

# Task 1: Increase by 15%, round to 2 decimals
increased = np.round(marks * 1.15, 2)
print("Task 1:", increased)

# Task 2: Cap at 100
capped = np.round(np.minimum(marks * 1.15, 100), 2)
print("\nTask 2:", capped)   # Expected: [ 46.  63.25 71.3  100. 100.  37.95 81.65 66.7 ]

# Task 3: How many students gained less than 10 marks (increase < 10)
print("\nTask 3:", np.sum((marks * 0.15) < 10))

# Task 4: Failing (< 40) before and after
print("\nTask 4 - Failing before:", np.sum(marks < 40), "| Failing after:", np.sum(capped < 40))

# Task 5: Each student as % of topper's ORIGINAL marks, rounded to 1 decimal
print("\nTask 5:", np.round(marks / marks.max() * 100, 1))

# Task 6: Total marks given away in revaluation
print("\nTask 6:", np.sum(capped - marks))
```

## Q10. Grace Marks and Weightage

```python
marks     = np.array([[72, 85, 60, 91],
                      [45, 30, 98, 55],
                      [88, 66, 74, 100],
                      [95, 92, 55, 40]])
grace     = np.array([5, 0, 8, 3])       # subject-wise (row broadcast)
weightage = np.array([1.2, 1.0, 1.1, 1.0])

# Task 1: Apply grace, cap at 100
graced = np.minimum(marks + grace, 100)
print("Task 1:\n", graced)

# Task 2: Multiply each subject by its weightage, round to 1 decimal
print("\nTask 2:\n", np.round(marks * weightage, 1))

# Task 3: Deviation from subject-wise class average
deviations = (marks - marks.mean(axis=0)).astype(float)
print("\nTask 3:\n", deviations)

# Task 4: Bonus per student as a column vector (5, 0, 2, 0)
bonus = np.array([5, 0, 2, 0]).reshape(-1, 1)
print("\nTask 4:\n", marks + bonus)

# Task 5: Count entries below subject average (negative deviations)
print("\nTask 5:", np.sum(deviations < 0))

# Task 6: Standardize each column (subtract mean, divide by std)
col_std  = marks.std(axis=0)
col_mean = marks.mean(axis=0)
standardized = np.round((marks - col_mean) / col_std, 2)
print("\nTask 6:\n", standardized)
```

## Q11. Distances and Arrangements

```python
metres   = np.array([200, 500, 900, 100, 700])
sanskrit = np.array([65, 70, 88, 92, 54])

# Task 1: 5x5 distance table (all positive) using broadcasting
dist = np.abs(metres[:, np.newaxis] - metres)
print("Task 1:\n", dist)

# Task 2: For each block, farthest distance
print("\nTask 2:", dist.max(axis=1))   # Expected: [700 400 800 800 600]

# Task 3: Block nearest to block 0 (exclude block 0 itself)
d0 = dist[0].copy().astype(float)
d0[0] = np.inf
print("\nTask 3: Block", np.argmin(d0))

# Task 4: Two blocks closest to each other
d_no_diag = dist.astype(float).copy()
np.fill_diagonal(d_no_diag, np.inf)
idx = np.unravel_index(np.argmin(d_no_diag), d_no_diag.shape)
print("\nTask 4: Closest pair of blocks:", idx)

# Task 5: Roll numbers 1-24 into rows of 6 (without typing row count)
rolls = np.arange(1, 25).reshape(-1, 6)
print("\nTask 5:\n", rolls)

# Task 6: sanskrit as column vector — two methods
col_v1 = sanskrit.reshape(-1, 1)
col_v2 = sanskrit[:, np.newaxis]
print("\nTask 6 (method 1 shape):", col_v1.shape, "| (method 2 shape):", col_v2.shape)

# Task 7: sanskrit as single-row table of shape (1, 5)
row_table = sanskrit.reshape(1, -1)
print("\nTask 7:", row_table, "| shape:", row_table.shape)
```

## Q12. The Master Mark Sheet

```python
master = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

# Task 1: Flat COPY — change first value to 999, master unchanged
flat_copy = master.flatten()
flat_copy[0] = 999
print("Task 1 - master (unchanged):\n", master)

# Task 2: Flat VIEW — change last value to 111, master changes
flat_view = master.ravel()
flat_view[-1] = 111
print("\nTask 2 - master (changed via ravel):\n", master)
# Restore master for remaining tasks
master = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

# Task 3: Set middle row to 0 without changing master
t3 = master.copy(); t3[1] = 0
print("\nTask 3 - copy with middle row zeroed:\n", t3)
print("master unchanged:\n", master)

# Task 4: Double first two columns without changing master
t4 = master.copy(); t4[:, :2] *= 2
print("\nTask 4:\n", t4)

# Task 5: Double first two columns — changing master itself
master[:, :2] *= 2
print("\nTask 5 - master (modified):\n", master)

# Task 6: Copy where every value > 50 reduced by 10
t6 = np.where(master > 50, master - 10, master)
print("\nTask 6 - copy (>50 reduced):\n", t6)
print("master unchanged:\n", master)
```

## Q13. Merging Section Records

```python
sec_A  = np.array([[72, 85, 60], [45, 30, 98], [88, 66, 74]])
sec_B  = np.array([[95, 92, 55], [30, 48, 62]])
roll_C  = np.array([201, 202, 203, 204])
marks_C = np.array([67, 82, 45, 91])

# Task 1: Combine A and B
combined = np.vstack([sec_A, sec_B])
print("Task 1 shape:", combined.shape)
print(combined)

# Task 2: Attach totals column
with_totals = np.hstack([combined, combined.sum(axis=1, keepdims=True)])
print("\nTask 2:\n", with_totals)

# Task 3: Two-column table from roll_C and marks_C
table_C = np.column_stack([roll_C, marks_C])
print("\nTask 3:\n", table_C)

# Task 4: Add new student [205, 73]
table_C_new = np.vstack([table_C, [205, 73]])
print("\nTask 4:\n", table_C_new)

# Task 5: Attach serial-number column (starting from 1) to combined
serial = np.arange(1, combined.shape[0] + 1).reshape(-1, 1)
print("\nTask 5:\n", np.hstack([serial, combined]))

# Task 6: Only students whose total > 200
totals = with_totals[:, -1]
print("\nTask 6:\n", combined[totals > 200])

# Task 7: Grade column: >=225 -> A, >=150 -> B, else C
grades = np.where(totals >= 225, 'A', np.where(totals >= 150, 'B', 'C'))
graded = np.column_stack([combined, grades])
print("\nTask 7:\n", graded)
```

## Q14. Cleaning and Grading

```python
marks      = np.array([[70, 65, 80],
                       [55, 90, 32],
                       [88, 76, 91],
                       [40, 52, 45],
                       [60, 58, 66]])
attendance = np.array([88, -1, 105, 76, 45, -1, 92, 120, 67, 30])

# Task 1: Repair attendance — cap >100 to 100, then fill -1 with mean of valid
temp       = np.minimum(attendance, 100)             # cap first
valid_mean = temp[attendance != -1].mean()           # mean of genuine entries
repaired   = np.where(attendance == -1, valid_mean, temp)
print("Task 1:", repaired)   # Expected: [ 88.  74.75 100.  76.  45.  74.75 92. 100.  67.  30.]

# Task 2: Label each repaired value
labels = np.where(repaired >= 85, 'Regular',
         np.where(repaired >= 60, 'Average', 'Short'))
print("\nTask 2:", labels)

# Task 3: Index positions and values of Short students
short_idx = np.where(repaired < 60)[0]
print("\nTask 3 - Short indices:", short_idx, "| values:", repaired[short_idx])

# Task 4: Row & column of every mark below 50 in marks table
rows, cols = np.where(marks < 50)
print("\nTask 4 - rows:", rows, "| cols:", cols)

# Task 5: Actual marks at those positions
print("\nTask 5:", marks[rows, cols])   # Expected: [32 40 45]

# Task 6: Replace marks < 35 with 35, report count changed
count_changed = np.sum(marks < 35)
marks_fixed = np.where(marks < 35, 35, marks)
print("\nTask 6 - entries changed:", count_changed)
print(marks_fixed)

# Task 7: Student with highest count of marks below 60
print("\nTask 7:", np.argmax(np.sum(marks < 60, axis=1)))   # Expected: 3
```

## Q15. The Final Weighted Result

```python
M = np.array([[70, 65, 80],
              [55, 90, 72],
              [88, 76, 91],
              [60, 58, 66]])
w = np.array([0.3, 0.3, 0.4])
W = np.array([[0.3, 0.5],
              [0.3, 0.2],
              [0.4, 0.3]])

# Task 1: Weighted final score using w  (dot product per student)
scores_w = M @ w
print("Task 1:", scores_w)

# Task 2: Index of student with highest weighted score
print("\nTask 2:", np.argmax(scores_w))

# Task 3: Subject-wise total using multiplication (ones vector), not .sum()
print("\nTask 3:", np.ones(M.shape[0]) @ M)   # Expected: [273. 289. 309.]

# Task 4: Both final scores for all students using W  -> shape (4, 2)
results = M @ W
print("\nTask 4:\n", results)   # Expected: [[72.5 72.],[72.3 67.1],[85.6 86.5],[61.8 61.4]]

# Task 5: How many students score higher under the second scheme
print("\nTask 5:", np.sum(results[:, 1] > results[:, 0]))

# Task 6: Difference (scheme2 - scheme1) per student, rounded to 2 decimals
print("\nTask 6:", np.round(results[:, 1] - results[:, 0], 2))   # Expected: [-0.5 -5.2  0.9 -0.4]

# Task 7: Attach better of the two scores as a 4th column to M
better = results.max(axis=1, keepdims=True)
M_with_best = np.hstack([M, better])
print("\nTask 7 shape:", M_with_best.shape)
print(M_with_best)
```
