# Python String Functions — Practice Assignment Solutions

Solutions use string methods and basic built-ins.

```python
# Q1 — The Strict Username Validator
def validate_username(username):
    clean = username.strip()
    return clean.lower() if len(clean) >= 5 and clean.isalnum() else 'Invalid'

print(validate_username(' CodeMaster99 '))
print(validate_username(' Python_Dev '))
```

```python
# Q2 — The Spy Code Reverser
def decode_secret_code(secret_code):
    return ' '.join(secret_code.split('-')[::-1]).upper()

print(decode_secret_code('charlie-bravo-alpha'))
```

```python
# Q3 — The Messy Database Cleaner
def clean_names(raw_names):
    names = [' '.join(name.split()).title() for name in raw_names.split(';')]
    return ',\n'.join(names)

raw_names = ' john doe ; jAnE sMiTh;\nALAN tUrInG '
print(clean_names(raw_names))
```

```python
# Q4 — The Multiple Space Eliminator
def clean_sentence(text):
    return ' '.join(text.split()).capitalize()

text = ' python is a very\npowerful language '
print(clean_sentence(text))
```

```python
# Q5 — The Palindrome Sentence Checker
def is_sentence_palindrome(sentence):
    clean = ''.join(character for character in sentence.lower() if character.isalnum())
    return clean == clean[::-1]

print(is_sentence_palindrome('Was it a car or a cat I saw?'))
print(is_sentence_palindrome('Python is fun'))
```

```python
# Q6 — The Hashtag Extractor
def extract_hashtags(post):
    return [word[1:].lower() for word in post.split() if word.startswith('#')]

post = 'Loving the SUNSET! #Nature\n#Sunset_Vibes #beautiful'
print(extract_hashtags(post))
```

```python
# Q7 — The File Extension Finder
def find_extensions(filenames):
    extensions = [filename.strip().rsplit('.', 1)[1].upper() for filename in filenames.split(',')]
    return list(dict.fromkeys(extensions))

filenames = 'report.pdf, image.JPG,\ndata.csv, notes.txt,\nphoto.jpg'
print(find_extensions(filenames))
```

```python
# Q8 — The Domain Name Slicer
def extract_domain(email):
    if '@' not in email:
        return 'Invalid'
    domain_part = email.split('@', 1)[1]
    return domain_part.split('.', 1)[0].title()

print(extract_domain('developer_99@microsoft.com'))
print(extract_domain('admin_at_website.org'))
```

```python
# Q9 — The Anagram Detective
def are_anagrams(str1, str2):
    clean1 = ''.join(str1.lower().split())
    clean2 = ''.join(str2.lower().split())
    return sorted(clean1) == sorted(clean2)

print(are_anagrams('Dormitory', 'Dirty room'))
```

```python
# Q10 — The Messy Phone Number Formatter
def format_phone_number(raw_num):
    digits = ''.join(character for character in raw_num if character.isdigit())
    return f'({digits[:3]}) {digits[3:6]}-{digits[6:]}' if len(digits) == 10 else 'Invalid Number'

print(format_phone_number(' +1 (555) 123.4567 '))
print(format_phone_number('555-987-6543'))
```

```python
# Q11 — The Product SKU Inspector
def inspect_sku(sku):
    clean = sku.strip().upper()
    valid = (len(clean) == 8 and clean[3] == '-' and
             clean[:3].isalpha() and clean[4:].isdigit())
    return f'Valid: {clean}' if valid else 'Invalid SKU'

print(inspect_sku(' xYz-9876 '))
print(inspect_sku('AB-12345'))
print(inspect_sku('A@C-1234'))
```
