## Question 1: The Strict Username Validator
Validate a username: must be ≥5 chars (after stripping), only letters and digits allowed.
Return lowercase username if valid, else `"Invalid"`.

```python
def validate_username(username):
    cleaned = username.strip()
    if len(cleaned) < 5:
        return "Invalid"
    if not cleaned.isalnum():
        return "Invalid"
    return cleaned.lower()

print(validate_username(" CodeMaster99 "))  # Expected: codemaster99
print(validate_username(" Python_Dev "))    # Expected: Invalid
```

## Question 2: The Spy Code Reverser
Reverse hyphen-separated words, convert to UPPERCASE, join with spaces.

```python
secret_code = "charlie-bravo-alpha"
lst = secret_code.split("-")
lst = lst[::-1]
lst = [i.upper() for i in lst]
ans = " ".join(lst)
print(ans)  # Expected: ALPHA BRAVO CHARLIE
```

## Question 3: The Messy Database Cleaner
Clean semicolon-separated names: Title Case, strip spaces, join with `", "`.

```python
def clean_names(raw_names):
    lst = raw_names.split(";")
    cleaned = [name.strip().title() for name in lst]
    return ", ".join(cleaned)

raw_names = "  john doe ;   jAnE sMiTh;  ALAN tUrInG "
print(clean_names(raw_names))  # Expected: John Doe, Jane Smith, Alan Turing
```

## Question 4: The Multiple Space Eliminator
Remove extra spaces (one space between words), capitalize only the first letter. No `re` module.

```python
text = "   python    is   a    very   powerful  language   "
words = text.split()
result = " ".join(words)
result = result.capitalize()
print(result)  # Expected: Python is a very powerful language
```

## Question 5: The Palindrome Sentence Checker
Remove non-alphanumeric chars, lowercase, check if forwards == backwards.

```python
def is_palindrome(sentence):
    cleaned = "".join(ch.lower() for ch in sentence if ch.isalnum())
    return cleaned == cleaned[::-1]

print(is_palindrome("Was it a car or a cat I saw?"))  # Expected: True
print(is_palindrome("Python is fun"))                  # Expected: False
```

## Question 6: The Hashtag Extractor
Extract all hashtags, return lowercase without `#`.

```python
post = "Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"
words = post.split()
hashtags = [word[1:].lower() for word in words if word.startswith("#")]
print(hashtags)  # Expected: ['nature', 'sunset_vibes', 'beautiful']
```

## Question 7: The File Extension Finder
Extract unique file extensions from comma-separated filenames, return in UPPERCASE.

```python
filenames = "report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"
files = filenames.split(", ")
seen = []
for f in files:
    ext = f.strip().split(".")[-1].upper()
    if ext not in seen:
        seen.append(ext)
print(seen)  # Expected: ['PDF', 'JPG', 'CSV', 'TXT']
```

## Question 8: The Domain Name Slicer
Extract domain name from email (between `@` and `.`), Title Case it. Return `"Invalid"` if no `@`.

```python
def extract_domain(email):
    if "@" not in email:
        return "Invalid"
    after_at = email.split("@")[1]
    domain = after_at.split(".")[0]
    return domain.title()

print(extract_domain("developer_99@microsoft.com"))  # Expected: Microsoft
print(extract_domain("admin_at_website.org"))         # Expected: Invalid
```

## Question 9: The Anagram Detective
Check if two strings are anagrams (ignore spaces, case-insensitive).

```python
def are_anagrams(str1, str2):
    clean1 = str1.replace(" ", "").lower()
    clean2 = str2.replace(" ", "").lower()
    return sorted(clean1) == sorted(clean2)

print(are_anagrams("Dormitory", "Dirty room"))  # Expected: True
```

## Question 10: The Messy Phone Number Formatter
Remove non-digit chars. If exactly 10 digits, format as `(XXX) XXX-XXXX`, else `"Invalid Number"`.

```python
def format_phone(raw_num):
    digits = "".join(ch for ch in raw_num if ch.isdigit())
    if len(digits) != 10:
        return "Invalid Number"
    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"

print(format_phone(" +1 (555) 123.4567 "))  # Expected: Invalid Number (11 digits)
print(format_phone("555-987-6543"))           # Expected: (555) 987-6543
```

## Question 11: The Product SKU Inspector
Validate SKU: exactly 3 uppercase letters + `-` + exactly 4 digits. No `re` module.

```python
def validate_sku(sku):
    cleaned = sku.strip().upper()
    if cleaned.count("-") != 1:
        return "Invalid SKU"
    parts = cleaned.split("-")
    letters_part = parts[0]
    digits_part  = parts[1]
    if len(letters_part) != 3 or not letters_part.isalpha():
        return "Invalid SKU"
    if len(digits_part) != 4 or not digits_part.isdigit():
        return "Invalid SKU"
    return f"Valid: {cleaned}"

print(validate_sku(" xYz-9876 "))  # Expected: Valid: XYZ-9876
print(validate_sku("AB-12345"))     # Expected: Invalid SKU
print(validate_sku("A@C-1234"))     # Expected: Invalid SKU
```

## Question 1: Selective Word Frequency Counter
Count word frequencies ignoring stop_words. Case-insensitive.

```python
def word_frequency(text, stop_words):
    stop_words = [w.lower() for w in stop_words]
    freq = {}
    for word in text.lower().split():
        if word not in stop_words:
            freq[word] = freq.get(word, 0) + 1
    return freq

text = "The quick brown fox jumps over the lazy dog"
stop_words = ["the", "over", "is", "a"]
print(word_frequency(text, stop_words))
# Expected: {'quick': 1, 'brown': 1, 'fox': 1, 'jumps': 1, 'lazy': 1, 'dog': 1}
```

## Question 2: The Inventory Merger
Merge two inventory dicts, summing quantities for shared items.

```python
branch_A = {"laptop": 10, "mouse": 50, "keyboard": 20}
branch_B = {"mouse": 30, "monitor": 15, "laptop": 5}

master = dict(branch_A)
for item, qty in branch_B.items():
    master[item] = master.get(item, 0) + qty

print(master)
# Expected: {'laptop': 15, 'mouse': 80, 'keyboard': 20, 'monitor': 15}
```

## Question 3: The Dictionary Inverter
Invert {student: grade} → {grade: [students]}.

```python
grades = {"Ravi": "A", "Neha": "B", "Aman": "A", "Pooja": "C", "Rahul": "B"}

inverted = {}
for student, grade in grades.items():
    if grade not in inverted:
        inverted[grade] = []
    inverted[grade].append(student)

print(inverted)
# Expected: {'A': ['Ravi', 'Aman'], 'B': ['Neha', 'Rahul'], 'C': ['Pooja']}
```

## Question 4: The Salary Bonus Filter
Filter employees earning < 60,000 and apply a 10% bonus.

```python
salaries = {"Aman": 50000, "Neha": 75000, "Rahul": 55000, "Pooja": 90000}

bonus_dict = {name: salary * 1.10
              for name, salary in salaries.items()
              if salary < 60000}

print(bonus_dict)
# Expected: {'Aman': 55000.0, 'Rahul': 60500.0}
```

## Question 5: The Safe API Extractor
Map user id → city using `.get()` for safe nested access (no try-except, no KeyError).

```python
def extract_cities(users):
    result = {}
    for user in users:
        uid = user["id"]
        address = user.get("profile", {}).get("address", {})
        city = address.get("city", "Unknown")
        result[uid] = city
    return result

users = [
    {"id": 101, "name": "Aman",  "profile": {"address": {"city": "Delhi"}}},
    {"id": 102, "name": "Neha",  "profile": {}},
    {"id": 103, "name": "Rahul", "profile": {"address": {"zip": "400001"}}}
]

print(extract_cities(users))
# Expected: {101: 'Delhi', 102: 'Unknown', 103: 'Unknown'}
```
