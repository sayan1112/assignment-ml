## Q4. Missing values

First, `info()` reports the non-null count for every column. The columns with missing values are:

## Q5. Why `describe()` shows fewer columns

`describe()` defaults to summarising numeric columns. For example, it leaves out `sex` and `embarked` because they are categorical/text columns rather than numeric measurement columns. Other non-numeric columns such as `class`, `who`, `deck`, `embark_town`, and `alive` are also left out.

```python
display(t.describe())
```

## Q6. The `class` column

Each question uses a different Pandas function:

- Different values: `nunique()` → **3**
- What they are: `unique()` → **First, Second, Third**
- Count in each group: `value_counts()` → **First: 216, Third: 491, Second: 184**

## Q8. Mean versus median age

The mean age is approximately **29.70**, while the 50% value (median) is **28.00**. The mean is larger, which suggests that the age distribution is **right-skewed**: some older passengers pull the mean upward.

```python
age_summary = t['age'].describe()
print('Mean age:', age_summary['mean'])
print('50% age:', age_summary['50%'])
print('Mean is larger:', age_summary['mean'] > age_summary['50%'])
```

## Q9. Percentage of missing age values

`shape` gives 891 total rows, and `info()` reports 714 non-null age values. Therefore:

`(891 - 714) / 891 × 100 = 19.87%`

So approximately **19.87% of age values are missing**.

```python
total_rows = t.shape[0]
age_non_null_from_info = 714  # t.info() reports 714 non-null age values
age_missing = total_rows - age_non_null_from_info
age_missing_percentage = age_missing / total_rows * 100
print(f'{age_missing_percentage:.2f}%')
```
